```python
import matplotlib.pyplot as plt

# Data from your Java output
data_sizes = [100, 500, 1000, 2500, 5000, 7500, 10000]
bubble_sort_times = [1.401, 9.557, 2.619, 14.290, 71.389, 193.664, 387.227]
merge_sort_times = [0.238, 0.447, 0.531, 0.720, 1.032, 1.212, 2.445]

# Linear Scale Plot (evenly spaced labels)
plt.figure(figsize=(10, 6))
plt.plot(range(len(data_sizes)), bubble_sort_times, marker='o', label='Bubble Sort', linewidth=2)
plt.plot(range(len(data_sizes)), merge_sort_times, marker='s', label='Merge Sort', linewidth=2)

plt.xticks(range(len(data_sizes)), data_sizes)  # evenly spaced data labels
plt.title("Execution Time Comparison: Bubble Sort vs Merge Sort (Linked List)")
plt.xlabel("Data Size (number of elements)")
plt.ylabel("Execution Time (ms)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Logarithmic Scale Plot (better for wide ranges)
plt.figure(figsize=(10, 6))
plt.plot(data_sizes, bubble_sort_times, marker='o', label='Bubble Sort', linewidth=2)
plt.plot(data_sizes, merge_sort_times, marker='s', label='Merge Sort', linewidth=2)

plt.xscale('log')  # logarithmic scale for x-axis
plt.title("Execution Time Comparison (Log Scale): Bubble Sort vs Merge Sort (Linked List)")
plt.xlabel("Data Size (log scale)")
plt.ylabel("Execution Time (ms)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```
